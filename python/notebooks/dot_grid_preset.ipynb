# Dot Grid Preset

```python
from femtum_sdk.api.contracts.v1.dto.galvo_pb2 import GalvoCalibrationEmpty, GalvoCalibrationSelector
from femtum_sdk.api.contracts.v1.dto.measure_pb2 import AngleMeasure, AngleMeasureUnit, DistanceMeasure, DistanceMeasureUnit, EnergyMeasure, EnergyMeasureUnit
from femtum_sdk.api.contracts.v1.hardware.laser_execution_pb2 import BurnDotGridRequest, LaserExecutionOptions

context_size_pixels = 4512
scaling_factor_mm_per_px = 0.0005376
context_size_microns = context_size_pixels * scaling_factor_mm_per_px * 1000

with HardwareApi(host=api_host) as hardware_api:
    hardware_api.laser_execution.BurnDotGrid(BurnDotGridRequest(
        Rows=5, 
        Columns=5, 
        SpacingX=DistanceMeasure(
            Value=200,
            Unit=DistanceMeasureUnit.Microns
        ), 
        SpacingY=DistanceMeasure(
            Value=200,
            Unit=DistanceMeasureUnit.Microns
        ), 
        OriginX=DistanceMeasure(
            Value=context_size_microns/2,
            Unit=DistanceMeasureUnit.Microns
        ), 
        OriginY=DistanceMeasure(
            Value=context_size_microns/2,
            Unit=DistanceMeasureUnit.Microns
        ),
        Rotation=AngleMeasure(
            Value=0,
            Unit=AngleMeasureUnit.Radians
        ),
        Options=LaserExecutionOptions(
            GalvoCalibrationSelector=GalvoCalibrationSelector(
                Empty=GalvoCalibrationEmpty()
            ),
            EnergyPerPulse=EnergyMeasure(
                Value=600, 
                Unit=EnergyMeasureUnit.Nanojoules
            )
        )
    ))
```
