# Execute Recipes Programatically without database

This notebook uses the Poetry-managed environment. All dependencies from `pyproject.toml` are available.


```python
# Verify we're using the correct environment
import sys
print(f"Python: {sys.executable}")
```

```python
from femtum_sdk.api.contracts.v1.instruments.health_pb2 import HealthRequest
from femtum_sdk.api import InstrumentApi

api_host = "localhost:5210"

with InstrumentApi(host=api_host) as instrument_api:
    instrument_health_response = instrument_api.health.GetHealth(HealthRequest())

    print("Instrument Health:")
    print(instrument_health_response)
```

## Execute Single Point Recipe

```python
from femtum_sdk.api.contracts.v1.dto.geometry_pb2 import Attribute, Geometry, Point, Shape
from femtum_sdk.api.contracts.v1.dto.measure_pb2 import EnergyMeasure, EnergyMeasureUnit
from femtum_sdk.api.contracts.v1.dto.recipe_pb2 import LaserParameter, Recipe
from femtum_sdk.api.contracts.v1.instruments.laser_execution_pb2 import ExecuteLaserExecutionRecipeRequest, ExecuteLaserExecutionRecipeResponse, LaserExecutionFeatures, LaserExecutionOptions

with InstrumentApi(host=api_host) as instrument_api:
    result: ExecuteLaserExecutionRecipeResponse = instrument_api.laser_execution.ExecuteRecipe(ExecuteLaserExecutionRecipeRequest(
        features=LaserExecutionFeatures(
            take_pictures=True,
        ),
        options=LaserExecutionOptions(
            # laser_energy_calibration_id="",
            energy_per_pulse=EnergyMeasure(
                value=0.5,
                unit=EnergyMeasureUnit.Nanojoules,
            ),
        ),
        # component_path="MyComponent",
        recipe=Recipe(
            geometries=[
                Geometry(
                    type="laser_marking",
                    attributes=[
                        Attribute(
                            key="marking_type",
                            value="single_pulse",
                        )
                    ],
                    shape=Shape(
                        point=Point(
                            x=100,
                            y=100,
                        )
                    ),
                )
            ],
            laser_parameters=[
                LaserParameter(key="single_pulse_laser_triggering_frequency_khz", value="2732", unit="kHz"),
            ]
        )))

    execution = result.execution
    print(execution)
```
