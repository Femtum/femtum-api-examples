## Execution Simulation

```python
from femtum_sdk.api.contracts.v1.dto.measure_pb2 import DistanceMeasure, DistanceMeasureUnit, EnergyMeasure, EnergyMeasureUnit
from femtum_sdk.api.contracts.v1.hardware.health_pb2 import HealthRequest
from femtum_sdk.api.contracts.v1.hardware.laser_execution_pb2 import LaserExecutionDistance, LaserExecutionOptions, SimulateLaserExecutionRecipeRequest, SimulateLaserExecutionRecipeResponse


simulation_iteration_1_response: SimulateLaserExecutionRecipeResponse = api.laser_execution.SimulateExecuteRecipe(SimulateLaserExecutionRecipeRequest(
    RecipeId="4",
    Options=LaserExecutionOptions(
        CalibrationId="3",
        Distance=LaserExecutionDistance(
            RelativeFromPreviousExecution=DistanceMeasure(
                Value=0.0,
                Unit=DistanceMeasureUnit.Microns,
            ),
        ),
        EnergyPerPulse=EnergyMeasure(
            Value=0.2,
            Unit=EnergyMeasureUnit.Nanojoules,
        ),
    ),
    PreviousExecutions=[],
))

simulation_iteration_2_response: SimulateLaserExecutionRecipeResponse = api.laser_execution.SimulateExecuteRecipe(SimulateLaserExecutionRecipeRequest(
    RecipeId="4",
    Options=LaserExecutionOptions(
        CalibrationId="3",
        Distance=LaserExecutionDistance(
            RelativeFromPreviousExecution=DistanceMeasure(
                Value=10.0,
                Unit=DistanceMeasureUnit.Microns,
            ),
        ),
        EnergyPerPulse=EnergyMeasure(
            Value=0.2,
            Unit=EnergyMeasureUnit.Nanojoules,
        ),
    ),
    PreviousExecutions=simulation_iteration_1_response.Executions,
))

print(simulation_iteration_2_response.Executions)

```
