```python
from femtum_sdk.api.contracts.v1.hardware.health_pb2 import HealthRequest
from femtum_sdk.api import HardwareApi

api = HardwareApi(host="localhost:5210")
api.__enter__()
health_response = api.health.GetHealth(HealthRequest())
print(health_response)
```

## Read Current Illumination

```python
from femtum_sdk.api.contracts.v1.dto.camera_pb2 import CameraIllumination
from femtum_sdk.api.contracts.v1.hardware.camera_pb2 import ReadIlluminationRequest

illumination: CameraIllumination = api.laser_head_camera.ReadIllumination(ReadIlluminationRequest())
print(illumination.PowerPercent)
```

## Update Current Illumination

```python
from femtum_sdk.api.contracts.v1.hardware.camera_pb2 import UpdateIlluminationRequest


api.laser_head_camera.UpdateIllumination(UpdateIlluminationRequest(Illumination=CameraIllumination(PowerPercent=50)))

updated_illumination: CameraIllumination = api.laser_head_camera.ReadIllumination(ReadIlluminationRequest())
print(updated_illumination.PowerPercent)
```
