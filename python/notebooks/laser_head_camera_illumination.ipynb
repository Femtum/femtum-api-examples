```python
from femtum_sdk.api.contracts.v1.instruments.health_pb2 import HealthRequest
from femtum_sdk.api import InstrumentApi

api_host = "localhost:5210"

with InstrumentApi(host=api_host) as api:
    health_response = api.health.GetHealth(HealthRequest())
    print(health_response)
```

## Read Current Illumination

```python
from femtum_sdk.api.contracts.v1.dto.camera_pb2 import CameraIllumination
from femtum_sdk.api.contracts.v1.instruments.camera_pb2 import ReadIlluminationRequest

with InstrumentApi(host=api_host) as api:
    illumination: CameraIllumination = api.laser_head_camera.ReadIllumination(ReadIlluminationRequest())
    print(illumination.power_percent)
```

## Update Current Illumination

```python
from femtum_sdk.api.contracts.v1.instruments.camera_pb2 import UpdateIlluminationRequest


with InstrumentApi(host=api_host) as api:
    api.laser_head_camera.UpdateIllumination(UpdateIlluminationRequest(illumination=CameraIllumination(power_percent=50)))

    updated_illumination: CameraIllumination = api.laser_head_camera.ReadIllumination(ReadIlluminationRequest())
    print(updated_illumination.power_percent)
```
